# Visualization with Seaborn and Matplotlib

**Question No:1**

```python
import seaborn as sns
#sns.set()
sns.set(style="darkgrid")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")
plt.rcParams['figure.figsize']=(10,10)
```

```python
# Read the data set
data_BM=pd.read_csv('/content/Titanic-Dataset.csv')
#drop the null values
data_BM=data_BM.dropna(how='any')
#Multiply item visibility by 100 to increase size
# Note: Please verify if 'Item_visibility' exists in the Titanic dataset, as it is typically a column from the BigMart Sales dataset.
if "Item_visibility" in data_BM.columns:
    data_BM["visibility scaled"]=data_BM["Item_visibility"]*100
# View the top results
data_BM.head()
```

```python
from matplotlib import figure
# Create countplot
plt.figure(figsize=(4,4))
sns.catplot(x='Embarked', kind="count", data=data_BM, hue='Sex')
plt.title('passenger count by Embarked port')
plt.legend(title='Gender')
plt.show()
```

*Observation: Southampton (s) had the highest number of passengers boarding the ship*

**Question No:2**

```python
# Create swarmplot
sns.catplot(x="Pclass", y="Age", kind='swarm', data=data_BM[:250], hue='Sex')
plt.title('Age Distribution across passenger classes')
plt.legend(title='Gender')
plt.show()
```

*Observation:* *The higest number of passengers of age group from 15 to 65 included in passenger class1*

**Question No:3**

```python
# Count male and female passengers
gender_counts = data_BM['Sex'].value_counts()

# Create pie chart
plt.figure(figsize=(6,6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', explode=(0.1,0), startangle=90)
plt.title('Gender Distribution of Titanic Passengers')
plt.show()
```

*Observation: Male passengers made up a larger proportion of the data set than female passengers.The exploded slice highlights the male passenger group.*

**Question No:4**

```python
# select the required columns with the correct capitalization
columns = ['Age', 'Fare', 'Pclass', 'SibSp', 'Parch', 'Survived']

# Calculate correlation using the correct DataFrame variable 'data_BM'
correlation = data_BM[columns].corr()

# Create heatmap
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Titanic Dataset')
plt.show()
```

*Observation: Passenger class and survival show a negative correlation. Fare and survival tend to have a positive correlation.*

**Question No:5**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Use the already loaded and available data_BM DataFrame from memory
df = data_BM

# Plot the violin plot (corrected 'sex' to 'Sex' to match the column name)
sns.violinplot(data=df, x='Pclass', y='Fare', hue='Sex', split=True, inner='quartile', palette='dark:magenta')
plt.title('Fare Distribution by Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.legend(title='Gender')
plt.show()
```

*Observation: First-class passengers generally paid higher fares than passengers in second and third class.*

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset using the already loaded data_BM from memory
df = data_BM

# Create a scatter relationship plot, coloring points by 'Survived'
sns.relplot(x="Age", y="Fare", hue="Survived", data=df, kind="scatter", palette="coolwarm")
plt.title('Age vs Fare by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

*Observation: The scatter plot shows the relationship between age and fare.The colours distinguish passengers who survivedfrom those who didnot.*
